# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image "given" in color mode
image = cv2.imread('given.jpg') 

# OpenCV loads in BGR, convert to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# Create translation matrix: 50 pixels right (x), 80 pixels down (y)
translation_matrix = np.float32([[1, 0, 50], [0, 1, 80]])

# Apply translation
# (width, height) of the image is passed as the output size
translated_image = cv2.warpAffine(image_rgb, translation_matrix, (image.shape[1], image.shape[0]))

```

```python
plt.figure(figsize=(10, 5))

# 1st subplot: Original Image
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis('off')

# 2nd subplot: Translated Image
plt.subplot(1, 2, 2)
plt.imshow(translated_image)
plt.title("Translated Image")
plt.axis('off')

plt.show()

```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the correct image
image = cv2.imread('baseball.jpg')

# Convert BGR to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
rows, cols = image_rgb.shape[:2]

# 1. Horizontal Shearing Matrix
# Ithu image ah side (right) ah izhukkum (slant aagum)
# Matrix: [[1, shear_x, 0], [0, 1, 0]]
M_horizontal = np.float32([[1, 0.5, 0], [0, 1, 0]])

# 2. Vertical Shearing Matrix
# Ithu image ah keezha (down) izhukkum (slant aagum)
# Matrix: [[1, 0, 0], [shear_y, 1, 0]]
M_vertical = np.float32([[1, 0, 0], [0.5, 1, 0]])
```

```python
# Apply horizontal shearing
horizontally_sheared = cv2.warpAffine(image_rgb, M_horizontal, (cols, rows))

# Apply vertical shearing
vertically_sheared = cv2.warpAffine(image_rgb, M_vertical, (cols, rows))
```

```python
plt.figure(figsize=(15, 5))

# 1st subplot: Original Image
plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title('Original Image')
plt.axis('off')

# 2nd subplot: Horizontally Sheared Image
plt.subplot(1, 3, 2)
plt.imshow(horizontally_sheared)
plt.title('Horizontally Sheared Image')
plt.axis('off')

# 3rd subplot: Vertically Sheared Image
plt.subplot(1, 3, 3)
plt.imshow(vertically_sheared)
plt.title('Vertically Sheared Image')
plt.axis('off')

# Adjust layout and display
plt.tight_layout()
plt.show()
```

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**  
2. Use `cv2.flip` with appropriate flip codes.
3. Display the following using a **2 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Reflected Image**  
   - **Vertically Reflected Image**  
   - **Both-axis Reflected Image**


```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('baseball.jpg')

# Convert BGR to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# 1. Horizontal Reflection (flip left <-> right)
# flipCode = 1
horizontal_flip = cv2.flip(image_rgb, 1)

# 2. Vertical Reflection (flip top <-> bottom)
# flipCode = 0
vertical_flip = cv2.flip(image_rgb, 0)

# 3. Both-axis Reflection (flip both horizontally and vertically)
# flipCode = -1
both_axis_flip = cv2.flip(image_rgb, -1)
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread('baseball.jpg')

# Convert BGR to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Get image dimensions (height, width)
height, width = image_rgb.shape[:2]
```

```python
# Calculate the center of the image
center = (width // 2, height // 2)

# Create a rotation matrix for 45 degrees
# (center, angle, scale)
rotation_matrix_45 = cv2.getRotationMatrix2D(center, 45, 1.0)

# Apply the rotation
rotated_45 = cv2.warpAffine(image_rgb, rotation_matrix_45, (width, height))
```

```python
# Create a rotation matrix for 90 degrees
rotation_matrix_90 = cv2.getRotationMatrix2D(center, 90, 1.0)

# Apply the rotation
rotated_90 = cv2.warpAffine(image_rgb, rotation_matrix_90, (width, height))
```

```python
plt.figure(figsize=(15, 5))

# 1st subplot: Original Image
plt.subplot(1, 3, 1)
plt.imshow(image_rgb)
plt.title('Original Image')
plt.axis('off')

# 2nd subplot: 45° Rotated Image
plt.subplot(1, 3, 2)
plt.imshow(rotated_45)
plt.title('45° Rotated Image')
plt.axis('off')

# 3rd subplot: 90° Rotated Image
plt.subplot(1, 3, 3)
plt.imshow(rotated_90)
plt.title('90° Rotated Image')
plt.axis('off')

# Adjust layout and display
plt.tight_layout()
plt.show()
```
